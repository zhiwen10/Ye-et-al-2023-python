# pipeline4_ephys

Python translation of `pipeline4_ephys.m` from Ye et al. 2023 (spirals): ephys prediction preprocessing pipeline. Regenerates every `ephys/*.mat` file consumed by `notebooks/figure4_ephys.ipynb` from the raw data release.

No figures are produced; progress is printed per session.

```python
# pipeline ephys
from pathlib import Path

import pandas as pd

DATA_FOLDER = Path(r"D:\data")

# load widefield + ephys session table
T = pd.read_csv(DATA_FOLDER / "tables" / "spirals_ephys_sessions_new2.csv")
```

## Figure 4g

```python
from spirals_py.ephys.preprocessing.getdVPrediction import getdVPrediction

save_folder = DATA_FOLDER / "ephys" / "dv_prediction"
# predict dV from spiking data, cross validated
getdVPrediction(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getdVPredictionPermute import getdVPredictionPermute

save_folder = DATA_FOLDER / "ephys" / "dv_permute"
# predict dV from spiking data, shuffled
getdVPredictionPermute(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getEphysROI import getEphysROI

save_folder = DATA_FOLDER / "ephys" / "roi"
# draw brain ROI (interactive polygon per session on first run; <fname>_roi.mat is cached and skipped afterwards)
getEphysROI(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getSpiralsRaw import getSpiralsRaw

save_folder = DATA_FOLDER / "ephys" / "spirals_raw"
# detect widefield spirals within the brain ROI
getSpiralsRaw(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getEphysSpiralsGrouping import getEphysSpiralsGrouping

save_folder = DATA_FOLDER / "ephys" / "spirals_raw_fftn"
# group spirals based on spatiotemporal structure
getEphysSpiralsGrouping(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getSpiralsPrediction import getSpiralsPrediction

save_folder = DATA_FOLDER / "ephys" / "spirals_predict"
# detect spirals in predicted data, no need to group
getSpiralsPrediction(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getSpiralsPredictionPermute import getSpiralsPredictionPermute

save_folder = DATA_FOLDER / "ephys" / "spirals_predict_permute"
# detect spirals in permuted predictions (no MATLAB counterpart; consumed by getSpiralComparePermute)
getSpiralsPredictionPermute(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getSpiralComparePredict import getSpiralComparePredict

save_folder = DATA_FOLDER / "ephys" / "spirals_compare"
# assess spiral pairs in raw and predicted data
getSpiralComparePredict(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.ephys.preprocessing.getSpiralComparePermute import getSpiralComparePermute

# assess spiral pairs in raw and permuted data
getSpiralComparePermute(T, DATA_FOLDER, save_folder)
```

## Figure 4h,i

```python
from spirals_py.ephys.preprocessing.getPhaseFlowMatchingIndex import getPhaseFlowMatchingIndex

save_folder = DATA_FOLDER / "ephys" / "flow_var"
# get matching index for phase and flow
getPhaseFlowMatchingIndex(T, DATA_FOLDER, save_folder)
```

## Extended Data Fig.13d

```python
from spirals_py.ephys.preprocessing.getVarOrderedByNeuron import getVarOrderedByNeuron

save_folder = DATA_FOLDER / "ephys" / "var_ordered"
# sort variance explained by neuron contribution
getVarOrderedByNeuron(T, DATA_FOLDER, save_folder)
```
