# pipeline2_axons

Python translation of `pipeline2_axons.m` from Ye et al. 2023 (spirals): axon-preference preprocessing pipeline. Regenerates every `axons/*.mat` and `revision/axons/*.csv` file consumed by `notebooks/figure2_axons.ipynb` from the raw data release.

No figures are produced; progress is printed per session.

```python
# pipeline axons
from pathlib import Path

import pandas as pd

DATA_FOLDER = Path(r"D:\data")

# load session table
T = pd.read_excel(DATA_FOLDER / "tables" / "spiralSessions3.xlsx")
```

## Figure 2c,e

```python
from spirals_py.axons.preprocessing.getAxonBiasTable import getAxonBiasTable

save_folder = DATA_FOLDER / "axons"
# extract soma and axon info for the 435 sensory neurons
getAxonBiasTable(DATA_FOLDER, save_folder)
```

## Figure 2d,f

```python
from spirals_py.axons.preprocessing.getSpiralsPhaseMap import getSpiralsPhaseMap

save_folder = DATA_FOLDER / "axons" / "spirals_70pixels_mean_flow_left"
# save spirals phase maps for all sessions
getSpiralsPhaseMap(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.axons.preprocessing.getSpiralsPhaseMapLeft import getSpiralsPhaseMapLeft

getSpiralsPhaseMapLeft(T, DATA_FOLDER, save_folder)
```

## axons revision

```python
from spirals_py.axons.preprocessing.getAxonBiasTableMO import getAxonBiasTableMO

save_folder = DATA_FOLDER / "revision" / "axons"
# get table for all cells in left hemisphere
getAxonBiasTableMO(DATA_FOLDER, save_folder)
```

```python
from spirals_py.axons.preprocessing.getAxonBiasTableMO2 import getAxonBiasTableMO2

# get table for all MO cells
getAxonBiasTableMO2(DATA_FOLDER, save_folder)
```

```python
from spirals_py.axons.preprocessing.getMOroi import getMOroi

# draw MO ROI (interactive polygon on first run; MO_roi.mat is cached)
getMOroi(DATA_FOLDER, save_folder)
```

```python
from spirals_py.axons.preprocessing.getAxonBiasTableSSp2 import getAxonBiasTableSSp2

getAxonBiasTableSSp2(DATA_FOLDER, save_folder)
```

```python
from spirals_py.axons.preprocessing.getSpiralsPhaseMap2 import getSpiralsPhaseMap2

save_folder = DATA_FOLDER / "axons" / "spirals_100pixels_mean_flow"
# save spirals phase maps for all sessions
getSpiralsPhaseMap2(T, DATA_FOLDER, save_folder)
```
