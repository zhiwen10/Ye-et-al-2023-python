# pipeline6_task

Python translation of `pipeline6_task.m` from Ye et al. 2023 (spirals): task preprocessing pipeline. Regenerates every `task/*.mat` file consumed by `notebooks/figure6_task.ipynb` from the raw data release.

No figures are produced; progress is printed per mouse/session.

```python
from pathlib import Path

from spirals_py.utils.paths import data_root, out_root

DATA_FOLDER = data_root()  # MATLAB data release (SPIRALS_DATA_ROOT overrides)
OUT_FOLDER = out_root()  # python outputs (SPIRALS_OUT_ROOT overrides)
```

## get psychometric curve and trial outcome

```python
from spirals_py.task.preprocessing.getPsychometricCurve import getPsychometricCurve

save_folder = OUT_FOLDER / "task" / "psychometric_curve"
# get psychometric curves for all subjects
getPsychometricCurve(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getTaskTrialOutcome import getTaskTrialOutcome

save_folder = OUT_FOLDER / "task" / "task_outcome"
# get task trial outcome for each trial
getTaskTrialOutcome(DATA_FOLDER, save_folder)
```

## get mean maps for different trial types

```python
from spirals_py.task.preprocessing.getMeanMapSession import getMeanMapSession

save_folder = OUT_FOLDER / "task" / "task_mean_maps"
# get mean map for all contrast and trial types across sessions
getMeanMapSession(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getMeanMapsAll import getMeanMapsAll

save_folder = OUT_FOLDER / "task" / "task_mean_maps"
# average mean maps at [-2,2]s around stim onset for 3 trial types
getMeanMapsAll(DATA_FOLDER, save_folder)
```

## get spirals for different trial types

```python
from spirals_py.task.preprocessing.getCorrectSpiralDensity import getCorrectSpiralDensity

save_folder = OUT_FOLDER / "task" / "spirals"
getCorrectSpiralDensity(DATA_FOLDER, save_folder)
```

## get phase and amplitude around stim onset in VISp

```python
from spirals_py.task.preprocessing.getTaskOnsetPhase import getTaskOnsetPhase

save_folder = OUT_FOLDER / "task" / "spirals"

freq1 = [0.05, 2]
# get phase and amplitude around onset time
getTaskOnsetPhase(DATA_FOLDER, save_folder, freq1)
```

```python
from spirals_py.task.preprocessing.getTaskOnsetPhase import getTaskOnsetPhase

save_folder = OUT_FOLDER / "task" / "spirals"

freq2 = [2, 8]
# get phase and amplitude around onset time
getTaskOnsetPhase(DATA_FOLDER, save_folder, freq2)
```

## spirals during each trial

```python
from spirals_py.task.preprocessing.getTaskSpirals import getTaskSpirals

save_folder = OUT_FOLDER / "task" / "spirals"
# task spirals at [-2,2]s around onset time in each trial
getTaskSpirals(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getPassiveSpirals import getPassiveSpirals

save_folder = OUT_FOLDER / "task" / "spirals"
# passive spirals at [-2,2]s around onset time in each trial
getPassiveSpirals(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getSprialCountByRadiusTime import getSprialCountByRadiusTime

save_folder = OUT_FOLDER / "task" / "spirals"
# sort task spirals by radius and time around stim onset
getSprialCountByRadiusTime(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getPassiveSpiralPrePost import getPassiveSpiralPrePost

save_folder = OUT_FOLDER / "task" / "spirals"
# concatenate all passive spirals before and after stim onset
getPassiveSpiralPrePost(DATA_FOLDER, save_folder)
```

```python
from spirals_py.task.preprocessing.getCorrectSpiralPrePost import getCorrectSpiralPrePost

save_folder = OUT_FOLDER / "task" / "spirals"
# concatenate all task spirals before and after stim onset
getCorrectSpiralPrePost(DATA_FOLDER, save_folder)
```
