# pipeline5_whisker

Python translation of `pipeline5_whisker.m` from Ye et al. 2023 (spirals): whisker-stimulus preprocessing pipeline. Regenerates every `whisker/*.mat` file consumed by `notebooks/figure5_whisker.ipynb` from the raw data release.

No figures are produced; progress is printed per mouse/session.

```python
from pathlib import Path

from spirals_py.utils.paths import data_root, out_root

DATA_FOLDER = data_root()  # MATLAB data release (SPIRALS_DATA_ROOT overrides)
OUT_FOLDER = out_root()  # python outputs (SPIRALS_OUT_ROOT overrides)
```

## whisker mean maps

```python
from spirals_py.whisker.preprocessing.getWhiskerMeanMaps import getWhiskerMeanMaps

save_folder = OUT_FOLDER / "whisker" / "whisker_mean_maps"
# get whisker mean maps across 5 mice
getWhiskerMeanMaps(DATA_FOLDER, save_folder)
```

```python
from spirals_py.whisker.preprocessing.getSpiralsWhiskerMeanMaps import getSpiralsWhiskerMeanMaps

save_folder = OUT_FOLDER / "whisker" / "whisker_mean_maps"
# detect spirals from mean maps (brain-mask ROI drawn interactively on first run and cached to file)
getSpiralsWhiskerMeanMaps(DATA_FOLDER, save_folder)
```

## spirals peri stim

```python
from spirals_py.whisker.preprocessing.getSpiralsPrePost import getSpiralsPrePost

save_folder = OUT_FOLDER / "whisker" / "spirals_peri_stim"
# concatenate spirals pre and post stimulus
getSpiralsPrePost(DATA_FOLDER, save_folder)
```

```python
from spirals_py.whisker.preprocessing.getSpiralsPeriStim import getSpiralsPeriStim

save_folder = OUT_FOLDER / "whisker" / "spirals_peri_stim"
# concatenate spirals over time across trials
getSpiralsPeriStim(DATA_FOLDER, save_folder)
```

## single trials

```python
from spirals_py.whisker.preprocessing.getWhiskerSingleTrials import getWhiskerSingleTrials

save_folder = OUT_FOLDER / "whisker" / "single_trials"
# single trial maps (mouse ZYE_0092)
getWhiskerSingleTrials(DATA_FOLDER, save_folder)
```
