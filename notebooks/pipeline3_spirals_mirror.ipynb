# pipeline3_spirals_mirror

Python translation of `pipeline3_spirals_mirror.m` from Ye et al. 2023 (spirals): mirror-symmetric spiral preprocessing pipeline. Regenerates every `spirals_mirror/*.mat` file consumed by `notebooks/figure3_sprials_mirror.ipynb` from the raw data release.

No figures are produced; progress is printed per session.

```python
from pathlib import Path

from spirals_py.utils.paths import out_root

import pandas as pd

DATA_FOLDER = Path(r"D:\\data")  # MATLAB data release (inputs, read-only)
OUT_FOLDER = out_root()  # python outputs (D:\\data_python)

# load session table
T = pd.read_excel(DATA_FOLDER / "tables" / "spiralSessions3.xlsx")
```

## Figure 3c

```python
from spirals_py.spirals_mirror.preprocessing.getReducedRankRegressionAP import getReducedRankRegressionAP

save_folder = OUT_FOLDER / "spirals_mirror" / "regression_ap"
# save regression coeffs, kernels and R2 from AP regression
getReducedRankRegressionAP(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.spirals_mirror.preprocessing.getReducedRankRegressionHEMI import getReducedRankRegressionHEMI

save_folder = OUT_FOLDER / "spirals_mirror" / "regression_hemi"
# save regression coeffs, kernels and R2 from hemi regression
getReducedRankRegressionHEMI(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.spirals_mirror.preprocessing.getExampleKernelHEMI import getExampleKernelHEMI

save_folder = OUT_FOLDER / "spirals_mirror" / "matching_index"
# save kernel maps for example 8 pixels, hemi
getExampleKernelHEMI(T, DATA_FOLDER, save_folder)
```

```python
from spirals_py.spirals_mirror.preprocessing.getAxonMapAP import getAxonMapAP

save_folder = OUT_FOLDER / "spirals_mirror" / "matching_index"
# get axon projection maps in the anterior cortex (MO)
getAxonMapAP(DATA_FOLDER, save_folder)
```
