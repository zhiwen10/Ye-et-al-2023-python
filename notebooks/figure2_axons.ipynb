# figure2_axons

Python translation of `figure2_axons.m` from Ye et al. 2023 (spirals).

- Fig. 2a-c: `plotAxonOrientation` — axon orientation bias of all sensory neurons
- Fig. 2d: `plotBiasHitogram` — bias histogram and permutation
- Fig. 2e-f: `plotAxonFlowMatch2` — mean optical flow and axon-flow matching index
- Extended Data Fig. 10a-d: `plotAxonOrientationMO5` — axon bias of all neurons in left hemisphere (relative to MOp)

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DATA_FOLDER = Path(r"D:\data")
FIGURE_FOLDER = DATA_FOLDER / "figures"

# load widefield session table
T = pd.read_excel(DATA_FOLDER / "tables" / "spiralSessions3.xlsx")
T
```

## Figure 2

```python
from spirals_py.axons.plots.plotAxonOrientation import plotAxonOrientation

save_folder = FIGURE_FOLDER / "Fig2"
save_folder.mkdir(parents=True, exist_ok=True)

# plot axon bias of all sensory neurons
h2abc = plotAxonOrientation(DATA_FOLDER, save_folder)
plt.show()
```

```python
from spirals_py.axons.plots.plotBiasHitogram import plotBiasHitogram

# plot bias histogram and permutation
h2d = plotBiasHitogram(DATA_FOLDER, save_folder)
plt.show()
```

```python
from spirals_py.axons.plots.plotAxonFlowMatch2 import plotAxonFlowMatch2

# plot mean optical flow and axon-flow matching index
h2df = plotAxonFlowMatch2(T, DATA_FOLDER, save_folder)
plt.show()
```
