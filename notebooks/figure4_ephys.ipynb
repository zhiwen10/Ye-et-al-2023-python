# figure4_ephys

Python translation of `figure4_ephys.m` from Ye et al. 2023 (spirals). Main Fig. 4 (spiral prediction from subcortical spiking) and Extended Data Figs. 12-13.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DATA_FOLDER = Path(r"D:\data")
FIGURE_FOLDER = DATA_FOLDER / "figures"

# load widefield + ephys session table
T = pd.read_csv(DATA_FOLDER / "tables" / "spirals_ephys_sessions_new2.csv")
T
```

```python
save_folder = FIGURE_FOLDER / "Fig4"
save_folder.mkdir(parents=True, exist_ok=True)
```

```python
from spirals_py.ephys.plots.plotProbeLocation import plotProbeLocation

# plot all probe locations in atlas space
plotProbeLocation(T, DATA_FOLDER, save_folder)
plt.show()
```
